```python
# 3.1 1D NumPy Arrays
```

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)
times_four = arr * 4
print(times_four)
```

```python
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
```

```python
min_index = np.argmin(arr)
print(min_index)
max_index = np.argmax(arr)
print(max_index)
# corresponding values from arr
print(arr[min_index])
print(arr[max_index])
# Yes, the values match
print(np.min(arr))
print(np.max(arr))
```

```python
greater_20 = arr[arr > 20]
print(greater_20)
```

```python
# 3.2 2D NumPy Arrays
```

```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])
# Print the shape
print(matrix.shape)
```

```python
print(matrix[:,2])
print(matrix[1:])
print(matrix * 2)
```

```python
print(np.mean(matrix, axis=0))
print(np.mean(matrix))
print(np.max(matrix, axis=1))
print(np.min(matrix, axis=0))
```

```python
flat_matrix = matrix.flatten()
print(flat_matrix)
print(flat_matrix.shape)
```

```python
# The new shape is (12,) because calling .flatten made all of the numbers in one single row.
```

```python
flat_min = np.argmin(matrix)
flat_max = np.argmax(matrix)
print(flat_min)
print(flat_max)
```

```python
min_coords = np.unravel_index(flat_min, matrix.shape)
max_coords = np.unravel_index(flat_max, matrix.shape)
print(min_coords)
print(max_coords)
```

```python
greater_than_15 = matrix[matrix > 15]
print(greater_than_15)
```

```python
sorted_array = np.sort(matrix, axis=1)
print(sorted_array)
```

```python
# 4 Prime Star Systems
```

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def contains_primes(arr):
    prime_rows = []
    for row in arr:
        for num in row:
            if num >= 2:
                is_prime = True
                for i in range(2, num):
                    if num % i == 0:
                        is_prime = False
                        break
                if is_prime:
                    prime_rows.append(row)
                    break
    return np.array(prime_rows)
print(contains_primes(arr))
```

```python
# 5 Let's Play Checkers!
```

```python
# 5.1
def all_zeros():
    return np.zeros((8, 8), dtype = int)
print(all_zeros())
```

```python
# 5.2
def alt_pattern():
    board = np.zeros((8, 8), dtype = int)
    board[0::2, 0::2] = 1
    return board
print(alt_pattern())
```

```python
# 5.3
def checkerboard():
    board = np.zeros((8, 8), dtype = int)
    board[0::2, 0::2] = 1
    board[1::2, 1::2] = 1
    return board
print(checkerboard())
```

```python
# 5.4
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype = int)
    board [0::2, 1::2] = 1
    board [1::2, 0::2] = 1
    return board
print(reverse_checkerboard())
```

```python
# 6 The Expanding Universe
def expansion(arr, num_spaces): 
    spaces = " " * num_spaces
    expanded = []
    for word in arr:
        expanded.append(spaces.join(word))
    return np.array(expanded)

universe = np.array(['galaxy', 'clusters'])
print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
# 7 Second-Brightest Star
def secondDimmest(stars):
    sorted_stars = np.sort(stars, axis=1)
    return sorted_stars[:, -2]
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
print(secondDimmest(stars))
```
